```python
# Import modules
import numpy as np
import pandas as pd
from geopy import distance

# Import PyGad
import pygad
```

```python
#constant data
maxCapacity = 150.0 # capacidad del vehiculo en litros
consumption = 7.0 # consumo del vehiculo en Km por litro
```

```python
# leer los datos del archivo Excel y almacenarlos en un DataFrame
df = pd.read_excel('ciudades_Bahia.xlsx')

# generate cities 
print('cities:')
print('| id |   lat   |   long  |   cost  |')
cities = []

prices = df.iloc[:, 3].values

# recorrer el DataFrame y agregar cada fila como una tupla a la lista de ciudades
for index, row in df.iterrows():
    city = index
    lat = row[1]
    long = row[2]
    C = row[3]
    cities.append((city, lat, long, C))
    print('|%4i|%9.4f|%9.4f|%9.4f|' % (city, lat, long, C))
```

```python
N = len(cities) # Numero de ciudades
#Matriz con distancias
distance_matrix = np.zeros((N, N))
for i, start in enumerate(cities):
    for j, end in enumerate(cities):
        coord1 = (start[1], start[2]) # coordenadas de latitud y longitud del primer punto
        coord2 = (end[1], end[2]) # coordenadas de latitud y longitud del segundo punto
        distance_matrix[i][j] = distance.distance(coord1, coord2).km # distancia entre los dos puntos en Km
distance_matrix
```

De momento usaremos un camino fijo

```python
path_0 = np.array([10,  1,  7,  3, 18,  2, 29, 20, 15, 14, 28, 23,  9, 13, 24, 16, 19, 25,  6,  0,  4, 26, 17, 22,  8, 11, 21,  5, 12, 27])
km_0 = 3130.1393740079825
```

```python
# calculation of the minimum cost for a path
sorted_prices = np.sort(prices)
def minimun_path_cost(km):
    liters = km / consumption
    liters_aux = liters
    cost = 0
    for i in range(len(sorted_prices)):
        if liters_aux > maxCapacity:
            cost += maxCapacity * sorted_prices[i]
            liters_aux -= maxCapacity
        else:
            cost += liters_aux * sorted_prices[i]
            return cost

min_cost_0 = minimun_path_cost(km_0)    
min_cost_0
```

```python
# prices of the path
def path_prices(path):
    prices_arr = np.zeros(len(path))
    for i in range(len(path)):
        prices_arr[i] = prices[path[i]]
    return prices_arr
    

prices_0 = path_prices(path_0)
prices_0
```

```python
path = path_0
path_price = prices_0
min_cost = min_cost_0

def cost_func(solution):
    
    tank = 0.0
    cost = 0.0
    for i in range(len(solution)): # len(solution) = nuber of genes = len(path)-1
        distance_km = distance_matrix[path[i]][path[i+1]]
        distance_liters = distance_km / consumption
        if solution[i] == False: # minimum refueling to reach the next point
            if (distance_liters > tank):
                cost += (distance_liters - tank) * path_price[i]
                tank += distance_liters - tank
                
        else: # refueling to get to the next cheapest point without exceeding the limit
            liters_aux = distance_liters
            if (i <= len(solution)-1):
                for j in range(i+1, len(solution)):
                    if (path_price[j] < path_price[j+1]):
                        liters_aux += distance_matrix[path[j]][path[j+1]] / consumption
                    else:
                        break
                if liters_aux + tank > 150.0:
                    liters_aux = 150.0 - tank

            if (liters_aux > tank):
                cost += (liters_aux - tank) * path_price[i]
                tank += liters_aux - tank
                
        tank -= distance_liters
                  
    return cost

unos = np.ones(28, dtype=int)
ceros = np.zeros(28, dtype=int)
print(cost_func(unos))
print(cost_func(ceros))
```

```python

def fitness_func(ga_instance, solution, solution_idx):
    
    cost = cost_func(solution)

    x = (cost - min_cost) / min_cost
    if x <= 0:
        fitness = 1
    else:
        fitness = np.exp(-10*x)
                  
    return fitness

print(fitness_func(unos,unos,unos))
print(fitness_func(ceros,ceros,ceros))
```

```python
fitness_function = fitness_func

num_generations = 1000
num_parents_mating = 2

sol_per_pop = 8
num_genes = len(path) - 1


parent_selection_type = "sss"
keep_parents = 1

crossover_type = "single_point"

mutation_type = "random"
mutation_percent_genes = 10
```

```python
N = len(path)
gene_space = [[0, 1] for _ in range(num_genes)]
initial_population = np.array([np.ones((num_genes)) for i in range(sol_per_pop)])
```

```python
ga_instance = pygad.GA(num_generations=num_generations,
                       num_parents_mating=num_parents_mating,
                       fitness_func=fitness_function,
                       sol_per_pop=sol_per_pop,
                       num_genes=num_genes,
                       parent_selection_type=parent_selection_type,
                       keep_parents=keep_parents,
                       crossover_type=crossover_type,
                       mutation_type=mutation_type,
                       mutation_percent_genes=mutation_percent_genes,
                       gene_space=gene_space,
                       initial_population=initial_population)
```
